```python
import numpy as np
import matplotlib.pyplot as plt
from tensorflow.keras.models import load_model
```

```python
# -------------------------------
# Load Saved Generator
# -------------------------------
generator = load_model("dcgan_generator.keras")
```

```python
latent_dim = 100
```

```python
# -------------------------------
# Generate Images
# -------------------------------
noise = np.random.normal(0,
                         1,
                         (25, latent_dim))
```

```python
generated_images = generator.predict(noise, verbose=0)
```

```python
generated_images = (generated_images + 1) / 2
```

```python
# -------------------------------
# Display Images
# -------------------------------
fig, axes = plt.subplots(5,5, figsize=(7,7))
```

```python
count = 0
```

```python
for i in range(5):
    for j in range(5):
        axes[i,j].imshow(generated_images[count,:,:,0],
                         cmap='gray')

        axes[i,j].axis("off")

        count += 1
```

```python
plt.show()
```

```python
# -------------------------------
# Save Images
# -------------------------------
for i in range(25):

    plt.imsave(f"generated_image_{i+1}.png",
               generated_images[i,:,:,0],
               cmap="gray")
```

```python
print("25 Images Saved Successfully.")
```

```python
# Save the trained generator model
generator.save("dcgan_generator.keras")
```
